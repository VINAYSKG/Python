```python
def sum(*args):
  total=0
  for i in args:
    total+=i
  print("sum",total)
```

```python
sum(5,2,3,4,6)
```

```python
#factorial condition
def factorial(a):
  if a==1:
    return 1
  return a*factorial(a-1)
```

```python
factorial(5)
```

```python
def even(a):
  return a%2==0
```

```python
even(3)
```

```python
#cal Filter function
list1=[9,12,14,20,23]
a1=filter(even,list1)
print(list(a1))
```

```python
list2=[9,10,11,12,24,76]
even=list(filter(lambda x:x%2==0,list2)) # filter function to find (even no) even ==
odd=list(filter(lambda x:x%2!=0,list2)) # filter function to find (odd no)odd !=
print(tuple(even))
print(tuple(odd))
```

```python
#reduce function
from functools import reduce
list3=[10,9,7]
a1=reduce(lambda x1,x2:x1*x2,list3)
print(a1)
```
