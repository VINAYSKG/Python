```python
def sum(*args):
  total=0
  for i in args:
    total+=i
  print("sum",total)
```

```python
sum(5,2,3,4,6)
```

```python
#factorial condition
def factorial(a):
  if a==1:
    return 1
  return a*factorial(a-1)
```

```python
factorial(5)
```
