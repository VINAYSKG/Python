```python
# Restaurant Billing System
# abstraction oops
from abc import ABC, abstractmethod
import random

class FoodItem(ABC):
    def __init__(self, name, price):
        self.name = name
        self.price = price

    @abstractmethod
    def prepare(self):
        pass

# polymorphism oops
class Pizza(FoodItem):
    def prepare(self):
        return f"🍕 Preparing {self.name}!"

class Burger(FoodItem):
    def prepare(self):
        return f"🍔 Preparing {self.name}!"

class Pasta(FoodItem):
    def prepare(self):
        return f"🍝 Preparing {self.name}!"

class Soup(FoodItem):
    def prepare(self):
        return f"🍲 Preparing {self.name}!"

# menu
if __name__ == "__main__":
    menu = {
        "1": Pizza("Pizza", 250),
        "2": Burger("Burger", 150),
        "3": Pasta("Pasta", 200),
        "4": Soup("Soup", 100)
    }

    order_list = []
    invoice_no = random.randint(1000, 9999)

    while True:
        print("\n🍴 Restaurant Billing System 🍴")
        print("1. View Menu")
        print("2. Show Bill & Exit")
        choice = input("Enter your choice: ")

        if choice == "1":
            print("\n--- Menu ---")
            for key, item in menu.items():
                print(f"{key}. {item.name} - ₹{item.price}")

            # menu number
            food_choice = input("Enter item number to order (or 0 to go back): ")
            if food_choice in menu:
                try:
                    qty = int(input(f"Enter quantity of {menu[food_choice].name}: "))
                    if qty <= 0:
                        print("❌ Quantity must be greater than 0. Try again.")
                        continue
                    order_list.append((menu[food_choice], qty))
                    print(menu[food_choice].prepare())
                    print(f"✅ {qty} x {menu[food_choice].name} added to your order")
                except ValueError:
                    print("❌ Please enter a valid number for quantity.")
            elif food_choice == "0":
                continue
            else:
                print("❌ Invalid menu choice!")

        elif choice == "2":
            print("\n--- Bill Details ---")
            if not order_list:
                print("No items ordered.")
            else:
                customer_name = input("Enter Customer Name: ")
                customer_phone = input("Enter Customer Phone: ")

                # payment method selection
                print("\nSelect Payment Method:")
                print("1. Cash")
                print("2. Card")
                print("3. Online Payment")
                print("4. UPI")
                print("5. Wallet")
                payment_choice = input("Enter payment method choice: ")

                payment_method = {
                    "1": "Cash",
                    "2": "Card",
                    "3": "Online Payment",
                    "4": "UPI",
                    "5": "Wallet"
                }.get(payment_choice, "Unknown")

                print("\n===== INVOICE =====")
                print(f"Invoice No : {invoice_no}")
                print(f"Customer   : {customer_name}")
                print(f"Phone      : {customer_phone}")
                print(f"Payment    : {payment_method}")  # payment method
                print("--------------------")

                total = 0
                for item, qty in order_list:
                    cost = item.price * qty
                    total += cost
                    print(f"{item.name} x {qty} = ₹{cost}")

                # discount option
                while True:
                    try:
                        discount_percentage = float(input("Enter discount percentage (0 for no discount): "))
                        if 0 <= discount_percentage <= 100:
                            break
                        else:
                            print("❌ Discount percentage must be between 0 and 100.")
                    except ValueError:
                        print("❌ Please enter a valid number for discount percentage.")

                discount_amount = total * (discount_percentage / 100)
                subtotal_after_discount = total - discount_amount

                gst = subtotal_after_discount * 0.05
                grand_total = subtotal_after_discount + gst

                print(f"--------------------")
                print(f"Subtotal     = ₹{total}")
                print(f"Discount     = ₹{discount_amount:.2f} ({discount_percentage:.2f}%)")
                print(f"Subtotal (after discount) = ₹{subtotal_after_discount:.2f}")
                print(f"GST (5%)     = ₹{gst:.2f}")
                print(f"Total Bill   = ₹{grand_total:.2f}")
                print("====================")
                print("Thank you for dining with us 🙏")
            break

        else:
            print("❌ Invalid choice! Please try again.")
```
