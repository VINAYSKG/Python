```python
#class & object
class calc:
  def add(self,a,b):
    self.a=a
    self.b=b
    return self.a+self.b
  def sub(self,a,b):
    self.a=a
    self.b=b
    return self.a-self.b
obj=calc()
print(obj.add(3,4))
print(obj.sub(3,4))
```

```python
#class & object
class circle:
  def area(self,a):
    self.a=a
    return self.a*3.14*self.a
  def para(self,b):
    self.b=b
    return self.b*3.14*2

class square:
  def area(self,c):
    self.c=c
    return self.c*self.c
  def para(self,d):
    self.d=d
    return self.d+self.d+self.d+self.d

obj=circle()
print(obj.area(2))
print(obj.para(2))

obj1=square()
print(obj1.area(2))
print(obj1.para(2))
```

```python
#Inheritance
class Animal: #(parent class)
  def sound (self):
    print('Its a animal class')

class dog(Animal): #(child class)
  def sound(self):
    print('Its dog class')

#obj Cal
d=dog()       #obj=j
d.sound()

a=Animal()
a.sound()
```

```python
class fruits:
  def taste(self,a):
    self.a=a
    print('taste is:',self.a)

class orange(fruits):
  def index(self, ):
    print('index is high')

O=orange()
O.taste('sour')
O.index()
```

```python
#Inheritance method overriding
class Animal: #(parent class)
  def sound (self):
    print('Its a animal class')

class dog(Animal): #(child class)
  def sound(self):
    super().sound()
    print('Its dog class')

#obj Cal
d=dog()       #obj=j
d.sound()

a=Animal()
a.sound()
```

```python
#Inheritance operator overloading
class calci:
    def add(self,a,b):
        self.a=a
        self.b=b
        return self.a+self.b

#object calling
obj=calci()
print(obj.add(2,3))
print(obj.add('Python','Java'))
```

```python
#multiple postional args with class or
#Run time polymorphism
class calci:
  def var(self,*a):
    return sum(a)

obj=calci()
print(obj.var(2))
print(obj.var(2,4,9))
print(obj.var(2,3,4,))
```
