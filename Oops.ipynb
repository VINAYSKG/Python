```python
#class & object
class circle:
  def area(self,a):
    self.a=a
    return self.a*3.14*self.a
  def para(self,b):
    self.b=b
    return self.b*3.14*2

class square:
  def area(self,c):
    self.c=c
    return self.c*self.c
  def para(self,d):
    self.d=d
    return self.d+self.d+self.d+self.d

obj=circle()
print(obj.area(2))
print(obj.para(2))

obj1=square()
print(obj1.area(2))
print(obj1.para(2))
```

```python
#Inheritance
class Animal: #(parent class)
  def sound (self):
    print('Its a animal class')

class dog(Animal): #(child class)
  def sound(self):
    print('Its dog class')

#obj Cal
d=dog()       #obj=j
d.sound()

a=Animal()
a.sound()
```

```python
#Inheritance method overriding
class Animal: #(parent class)
  def sound (self):
    print('Its a animal class')

class dog(Animal): #(child class)
  def sound(self):
    super().sound()
    print('Its dog class')

#obj Cal
d=dog()       #obj=j
d.sound()

a=Animal()
a.sound()
```

```python
#multiple postional args with class or
#Run time polymorphism
class calci:
  def var(self,*a):
    return sum(a)

obj=calci()
print(obj.var(2))
print(obj.var(2,4,9))
print(obj.var(2,3,4,))
```

```python
#Inheritance method overriding
class circle1: #(parent class)
  def area (self,a):
    self.a=a
    return self.a*3.14*self.a

class squa(circle1): #(child class)
  def area(self,a):
    print(super().area(5))
    return self.a*self.a

obj= squa()
print(obj.area(2))     #obj=j
```

```python
#Inheritance single with constructor
class person:
  def __init__(self,name,id):
    self.name=name
    self.id=id

  def detail(self):
    print('name:', self.name,'id:',self.id)

class employee(person):
  def __init__(self,name,id,dept,sal):
    super().__init__(name,id)
    self.dept=dept
    self.sal=sal
  def details(self):
      return f"name:{self.name},id:{self.id},dept:{self.dept},sal:{self.sal}"

#obj calling
obj=employee('Riya',309,'sale',4800)
print(obj.details())
obj.detail()
```

```python
#multiple inheritance
class person:
  def detail_p(self):
    print('Its a person class')

class employee:
  def detail_e(self):
    print('Its employee class')

class player(person,employee):
  pass

#obj calling
obj=player()
obj.detail_e()
obj.detail_p()

print(player.__mro__) #for find multiple inheritance or parent child
```
